```python
import pandas as pd
import numpy as np
from scipy import stats

url_link_raw = "https://raw.githubusercontent.com/UsmanNaf/stat-audit-FinalProject_Group10-sti-2025/refs/heads/main/data/clean/pr_clean.csv"

df_pr = pd.read_csv(url_link_raw)
print(f"Data berhasil dimuat. Total baris data: {len(df_pr)}")

comments_merged = df_pr[df_pr['status'] == 'merged']['comments']
comments_unmerged = df_pr[df_pr['status'] == 'unmerged']['comments']

x_bar1, n1, sigma1 = comments_merged.mean(), len(comments_merged), comments_merged.std()
x_bar2, n2, sigma2 = comments_unmerged.mean(), len(comments_unmerged), comments_unmerged.std()

pooled_se = np.sqrt((sigma1**2 / n1) + (sigma2**2 / n2))
z_stat = (x_bar1 - x_bar2) / pooled_se
p_value = 2 * (1 - stats.norm.cdf(np.abs(z_stat)))

print("\n" + "=" * 55)
print("HASIL OUTPUT PENGUJIAN Z-TEST KELOMPOK")
print("=" * 55)
print(f"PR Merged   (Populasi 1) -> Mean: {x_bar1:.4f}, n: {n1}")
print(f"PR Unmerged (Populasi 2) -> Mean: {x_bar2:.4f}, n: {n2}")
print("-" * 55)
print(f"Z-Statistic Value : {z_stat:.4f}")
print(f"P-Value Result    : {p_value:.4e}")
print(f"Statistical Dec   : {'Reject H0' if p_value <= 0.05 else 'Fail to reject H0'}")
print("=" * 55)
```

## Prosedur 6 Langkah Uji Hipotesis

1. Hipotesis: H0 (Rata-rata komentar PR merged dan unmerged sama); Ha (Rata-rata komentar berbeda).
2. Alpha: 5% (0.05).
3. Nilai Statistik Uji: Berdasarkan hitungan diperoleh nilai Z sebesar -4.0185.
4. P-Value: Nilai p-value sebesar 5.85e-05 (sangat kecil mendekati nol).
5. Keputusan: Karena p-value <= 0.05, maka keputusannya adalah **Reject H0**.
6. Interpretasi: Berarti ada perbedaan nyata. PR yang ditolak (unmerged) komentarnya jauh lebih banyak karena banyak perdebatan atau review yang alot dari para maintainer GitHub.
